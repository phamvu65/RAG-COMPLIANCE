```python
# Cell 1 — Import
import pm4py
import pandas as pd

# Cell 2 — Đọc file XES
log = pm4py.read_xes('../data/raw/BPI_2017.xes')
print(f"Số case: {len(log)}")
print(f"Số event: {sum(len(trace) for trace in log)}")


# # Cell 3 — Chuyển sang DataFrame để dễ xem
df = pm4py.convert_to_dataframe(log)
print(df.columns.tolist())
df.head(10)

# Cell 4 — Xem danh sách activity có trong dataset
activities = df['concept:name'].unique()
print(f"Số loại activity: {len(activities)}")
for a in sorted(activities):
    print(f"  {a}")

# Cell 5 — Thống kê cơ bản
print("Số case:", df['case:concept:name'].nunique())
print("Khoảng thời gian:", df['time:timestamp'].min(), "→", df['time:timestamp'].max())
print("\nTop 10 activity phổ biến nhất:")
print(df['concept:name'].value_counts().head(10))

# Cell 6
import sys
import os
sys.path.append('../')
from src.event_log.preprocessor import load_and_clean, filter_complete_cases

df_clean = load_and_clean('../data/raw/BPI_2017.xes')
df_final = filter_complete_cases(df_clean, min_events=3, max_events=50)

# Tạo thư mục nếu chưa có
os.makedirs('../data/processed', exist_ok=True)

df_final.to_csv('../data/processed/event_log_clean.csv', index=False)
print("Đã lưu dataset sạch")
```
